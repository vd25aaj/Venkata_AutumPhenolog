# Import libraries

```python
# Libraries for downloading and reading the dataset
import io
import requests

# Libraries for data handling
import numpy as np
import pandas as pd

# Libraries for plots in later sections
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

# Load the original PhenoCam dataset

```python
# Original dataset URL from config.yml
DATA_URL = (
    "https://phenocam.nau.edu/data/archive/"
    "harvard/ROI/harvard_DB_1000_1day.csv"
)

# Download the CSV file
response = requests.get(DATA_URL, timeout=60)

# Stop and show an error if the download fails
response.raise_for_status()

# Remove metadata comments and blank lines
csv_text = "\n".join(
    line
    for line in response.text.splitlines()
    if line.strip() and not line.lstrip().startswith("#")
)

# Read the data into a DataFrame
df_raw = pd.read_csv(io.StringIO(csv_text))

# Remove accidental spaces from column names
df_raw.columns = df_raw.columns.str.strip()

# Check that the columns needed by the starter code are present
required_columns = [
    "date", "r_mean", "g_mean",
    "b_mean", "rcc_mean", "gcc_90"
]

missing_columns = [
    column for column in required_columns
    if column not in df_raw.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}. "
        "Check that the URL returned the correct CSV."
    )

print("Dataset loaded successfully.")
print("Number of rows:", df_raw.shape[0])
print("Number of columns:", df_raw.shape[1])

display(df_raw.head())
```

# Inspect the dataset

```python
# Show all column names
print("COLUMN NAMES")
print(df_raw.columns.tolist())

# Show data types and non-missing counts
print("\nDATASET INFORMATION")
df_raw.info()

# Count ordinary missing values
print("\nMISSING VALUES")
print(df_raw.isna().sum())

# Count the supervisor's missing-value marker
# Check both numeric and text representations
marker_counts = df_raw.apply(
    lambda column: column.astype(str).str.strip().isin(
        ["-9999", "-9999.0"]
    ).sum()
)

print("\nMISSING-VALUE MARKERS (-9999)")
print(marker_counts[marker_counts > 0])

# Count completely duplicated rows
print("\nDuplicate rows:", df_raw.duplicated().sum())

# Inspect dates without changing the original data
dates = pd.to_datetime(df_raw["date"], errors="coerce")

print("\nDATE COVERAGE")
print("First date:", dates.min())
print("Last date:", dates.max())
print("Invalid dates:", dates.isna().sum())

print("\nNUMBER OF OBSERVATIONS PER YEAR")
print(dates.dt.year.value_counts().sort_index())
```
